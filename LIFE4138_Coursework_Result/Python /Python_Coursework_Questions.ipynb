### Question 3

Use three different methods to print the numbers 200 to 300.

````python
# following are the different ways to print numbers between 200 to 300

# 1. using while loop
i = 200
while i <= 300:
    print(i)
    i += 1  

# 2. list comprehension
i = 200
j = 300

for num in range(i, j + 1):
    print(num)

# 3. For loop
for a in range(200, 301):
    print(a)
````

### Question 7
Analyze the life expectancy dataset (which can be loaded by running the cell below).

- Calculate the number of unique countries in the dataset.

- Calculate the country and year with the highest life expectancy.

- Calculate the year with the highest total expenditure.

- Provide your code using pandas.


````python
import seaborn as sns
life_expectancy = sns.load_dataset('healthexp')
````

````python
import pandas as pd
#1. Calculating the number of unique countries in the dataset
print(life_expectancy['Country'].nunique())
````

````python
#2. Calculating the country and year with the highest life expectancy
print(life_expectancy['Life_Expectancy'].max())
life_expectancy.loc[life_expectancy['Life_Expectancy'].idxmax(), ['Country', 'Year', 'Life_Expectancy']]
````

````python
#3. Calculating the year with highest total expenditure
yearly_expenditure = life_expectancy.groupby('Year')['Spending_USD'].sum()
highest_expenditure = yearly_expenditure.idxmax()
print(highest_expenditure) #year
print(yearly_expenditure.max()) #expense
````

### Question 8
Using the diamonds dataset and the plotting code of your choice, plot the following from the diamonds dataset.

1. Carat vs Price in a scatter plot.
2. A histogram of prices.
3. A boxplot of cut vs price.

The code can be loaded from seaborn in the cell below.

````python
diamonds = sns.load_dataset('diamonds')
diamonds.head()
````

````python
import pandas as pd
import matplotlib.pyplot as plt

#1. creating the scatterplot of 'carats' vs. 'price'
diamonds.plot.scatter(x='carat', y='price', s=5, alpha=0.5, color='purple')
plt.title('carat vs price')
plt.xlabel('carat')
plt.ylabel('price')
plt.show()
````

````python
#2. Creating a histogram of prices
plt.hist(diamonds['price'], bins=15, color='pink')
plt.title('Histogram of Fish Weight')
plt.xlabel('Prices')
plt.show()
````

````python
#3. Creating a boxplot of cut vs price
sns.boxplot(x='cut', y='price', data=diamonds)
plt.title('cut vs price')
plt.show()
````
