- the tallest height
- the lowest mass

HINT: the starwars dataset is part of the dplyr package

```r
#loading and viewing the starwars dataset
data(starwars)
library(tidyverse)

#finding the individual with the tallest height
starwars %>%
    arrange(desc(height)) %>%
    slice(1) %>%
    select(name, height)

#finding character with the lowest mass
starwars %>%
    arrange(mass) %>%
    slice(1) %>%
    select(name, mass)
```

### Question 7

The `dplyr` dataset `storms` is part of the US National Oceanic and Atmospheric Administration database of Atlantic hurricanes. It contains multiple repeated measurements from hurricanes. For a given storm, there are measurements from every 6 hours across the life time of the storm.

Use a tidyverse approach with `dplyr` to answer the following:

- how many different storms are contained in the database?
- which year had the most storms?
- which storm, in which year had the highest pressure?

HINT:
unique() will help ensure you remove duplicated data from a vector e.g.
x <- c(1, 1, 2, 2, 3, 3, 4, 4)
unique(x)

```r
#loading and viewing the storms dataset
data(storms)
library(dplyr)

#Total number of different storms in the database
storms %>%
    distinct(name) %>%
    tally()

#year with most storms
storms %>%
    distinct(name, year) %>%
    count(year, sort = TRUE) %>%
    top_n(1,n)
    
#storm with highest pressure 
storms %>%
    filter(!is.na(pressure)) %>%
    arrange(desc(pressure)) %>%
    select(name, year, pressure) %>%
    head(1)
```

### Question 8

The `ggplot2` dataset `diamonds` contains the price and information from the sale of 54,000 different diamonds. You can use ?diamonds to find out more about the dataset if you want.

Using `ggplot2`:

- plot the relationship between the carat (i.e. the weight of the diamond) and the price.
- plot a histogram of prices of all the diamonds
- plot a boxplot of the difference in price between the different cuts

```r
library(ggplot2)
#plot for relationship between carat and price 
ggplot(aes(x = carat, y = price), data = diamonds)+
    geom_point(size = 0.8, shape =22, alpha = 0.5, fill = "blue")

#histogram of prices
ggplot(aes(x = price), data = diamonds)+
    geom_histogram(binwidth = 300, fill="light pink", color="black")

#boxlot of difference of prices between different cuts
ggplot(aes(x = cut, y = price), data = diamonds)+
    geom_boxplot()
```

### Question 9

Identify the maximum and minimum price that the diamonds in the diamonds data set sold for. Add a new column to the diamonds data set that contains the string “Low price” if the value in the price column is less than 5000 dollars and “High price” if it is above 5000 dollars.

### Question 10

Sticking with the `diamonds` dataset for this question. A carat is a special unit of weight for gemstones where a single carat is 200 mg or 0.2 g. Write a function to convert the `diamonds$carat` vector into grams or milligrams. How heavy is the heaviest diamond?

```r
library(ggplot2)
data(diamonds)

#converting carats to grams or milligrams
carats_to <- function(carat_vector, unit = c("grams", "milligrams")) {
    unit <- match.arg(unit)
    if (unit == "grams") {
        conversion_factor <- 0.2
    } else {
        conversion_factor <- 200
    }
    return(carat_vector * conversion_factor)
}

#converting carat to gram
grams_vector <- carats_to(diamonds$carat, unit = "grams")

#converting carat to milligram
milligrams_vector <- carats_to(diamonds$carat, unit = "milligrams")

#adding new columns to dataset
diamonds$weight_g <- grams_vector
diamonds$weight_mg <- milligrams_vector

head(diamonds)

#heavy diamond
max_carat <- max(diamonds$carat)
print(paste("Max carats:", max_carat))

weight_grams <- carats_to(max_carat, unit = "grams")
print(paste("Max weight in grams:", weight_grams))

weight_milligrams <- carats_to(max_carat, unit = "milligrams")
print(paste("Max weight in milligrams:", weight_milligrams))
```
